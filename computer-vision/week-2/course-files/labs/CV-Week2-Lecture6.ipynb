## Why Face Detection?
Face detection is a fundamental step in many computer vision applications such as biometrics, surveillance, and human-computer interaction.

## Haar Cascades
- Haar features are simple rectangular features.
- Trained using AdaBoost to form a cascade classifier.
- Pros: Fast, lightweight.
- Cons: Sensitive to lighting and pose.

## HOG (Histogram of Oriented Gradients)
- Captures gradient orientation distributions.
- Used in classical face and pedestrian detection.
- Pros: Robust to illumination changes.
- Cons: Computationally heavier than Haar.

```python
import cv2

# Load Haar cascade
face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + 'haarcascade_frontalface_default.xml'
)

# Load image
img = cv2.imread('basketball.jpg')
if img is None:
    raise IOError("Image not found. Check the file path.")

# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Detect faces
faces = face_cascade.detectMultiScale(gray, scaleFactor=1.3, minNeighbors=5)

# Draw rectangles around detected faces
for (x, y, w, h) in faces:
    cv2.rectangle(img, (x, y), (x + w, y + h), (255, 0, 0), 2)

# Show result
cv2.imshow('Face Detection', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## Exercise: Detect Faces in Webcam Feed
Run the following code to detect faces live from your webcam.

```python
cap = cv2.VideoCapture(0)
while True:
    ret, frame = cap.read()
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    faces = face_cascade.detectMultiScale(gray, 1.3, 5)
    for (x,y,w,h) in faces:
        cv2.rectangle(frame,(x,y),(x+w,y+h),(255,0,0),2)
    cv2.imshow('Webcam Face Detection', frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
cap.release()
cv2.destroyAllWindows()
```

## Mini Project: Face Detector App
We will build a simple app that detects faces in images or webcam feed.

```python
def detect_faces(img_path=None, use_webcam=False):
    if use_webcam:
        cap = cv2.VideoCapture(0)
        while True:
            ret, frame = cap.read()
            gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
            faces = face_cascade.detectMultiScale(gray, 1.3, 5)
            for (x,y,w,h) in faces:
                cv2.rectangle(frame,(x,y),(x+w,y+h),(255,0,0),2)
            cv2.imshow('Face Detector App', frame)
            if cv2.waitKey(1) & 0xFF == ord('q'):
                break
        cap.release()
        cv2.destroyAllWindows()
    else:
        img = cv2.imread(img_path)
        gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
        faces = face_cascade.detectMultiScale(gray, 1.3, 5)
        for (x,y,w,h) in faces:
            cv2.rectangle(img,(x,y),(x+w,y+h),(255,0,0),2)
        cv2.imshow('Face Detector App', img)
        cv2.waitKey(0)
        cv2.destroyAllWindows()

# Example usage:
# detect_faces('face_sample.jpg')
# detect_faces(use_webcam=True)
```
