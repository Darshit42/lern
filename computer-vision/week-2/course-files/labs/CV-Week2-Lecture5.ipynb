# Day 12: Bag of Visual Words (BoVW)
Today we will cover:
- Feature Clustering
- Histogram Representation

## Why BoVW?
Bag of Visual Words (BoVW) is inspired by text analysis. Just like words in a document, local image features can be clustered into 'visual words'. Images are then represented as histograms of these words, enabling classification.

## Theory: Feature Clustering
- Extract local features (e.g., SIFT, ORB).
- Cluster descriptors using k-means.
- Each cluster center = a 'visual word'.

## Theory: Histogram Representation
- Assign each feature to nearest cluster.
- Build histogram of word occurrences.
- Use histogram as input to classifier.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report

# Example: Load texture images (replace with your dataset)
textures = ['sample.jpg','left.jpg','right.jpg']
labels = [0,1,2]  # numeric labels for classes

descriptors = []
y = []
sift = cv2.SIFT_create()

for idx, img_path in enumerate(textures):
    img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
    kp, des = sift.detectAndCompute(img, None)
    if des is not None:
        descriptors.extend(des)
        y.extend([idx]*len(des))

descriptors = np.array(descriptors)
print("Collected descriptors shape:", descriptors.shape)
```

## Exercise: Build BoVW Model for Textures
Cluster descriptors into visual words and represent each image as a histogram.

```python
# Cluster descriptors using k-means
k = 20  # number of visual words
kmeans = KMeans(n_clusters=k, random_state=42)
kmeans.fit(descriptors)

def build_histogram(img_path):
    img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
    kp, des = sift.detectAndCompute(img, None)
    hist = np.zeros(k)
    if des is not None:
        words = kmeans.predict(des)
        for w in words:
            hist[w] += 1
    return hist

# Build histograms for dataset
X = []
y = []
for idx, img_path in enumerate(textures):
    hist = build_histogram(img_path)
    X.append(hist)
    y.append(idx)

X = np.array(X)
y = np.array(y)
print("Feature matrix shape:", X.shape)
```

## Mini Project: Texture Classifier
Train an SVM classifier on BoVW histograms to classify textures.

```python
# Split into train/test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Train SVM
svm = SVC(kernel='linear')
svm.fit(X_train, y_train)
y_pred = svm.predict(X_test)

print("Classification Report:")
print(classification_report(y_test, y_pred))
```
