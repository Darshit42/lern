## Theory: Object Detection Architectures

### R-CNN (2014)
- Region-based CNN.
- Generates region proposals, then classifies each region.
- Accurate but slow.

### YOLO (2016)
- "You Only Look Once".
- Single-shot detector: predicts bounding boxes and classes in one pass.
- Real-time performance.

### SSD (2016)
- Single Shot MultiBox Detector.
- Uses multiple feature maps for detection at different scales.
- Faster than R-CNN, competitive accuracy.

## Setup: COCO Dataset Loader

```python
import torch
import torchvision
import torchvision.transforms as transforms
from torchvision.datasets import CocoDetection

# Example COCO dataset path (replace with actual path)
coco_root = './coco_data'

transform = transforms.Compose([
    transforms.Resize((224,224)),
    transforms.ToTensor()
])

trainset = CocoDetection(root=coco_root+'/train2017', annFile=coco_root+'/annotations/instances_train2017.json', transform=transform)
trainloader = torch.utils.data.DataLoader(trainset, batch_size=4, shuffle=True)

print("Loaded COCO dataset with", len(trainset), "images")
```

## Exercise: Detect Objects in COCO Dataset using Pretrained Model

```python
import torchvision.models.detection as detection

# Load pretrained Faster R-CNN model
model = detection.fasterrcnn_resnet50_fpn(pretrained=True)
model.eval()

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)

# Run detection on a batch
images, targets = next(iter(trainloader))
images = [img.to(device) for img in images]
outputs = model(images)

print("Detected objects in batch:")
for i, output in enumerate(outputs):
    print(f"Image {i}: {len(output['boxes'])} objects detected")
```

## Mini Project: YOLO-based Object Detector

```python
import torch
import cv2
from matplotlib import pyplot as plt

# Example: Using YOLOv5 from ultralytics (requires installation)
# pip install ultralytics
from ultralytics import YOLO

# Load YOLOv5 pretrained model
yolo_model = YOLO('yolov5s.pt')

# Run detection on a sample image
sample_image = 'sample.jpg'  # replace with actual image path
results = yolo_model(sample_image)

# Display results
results.show()

# Print detected classes
for r in results:
    print(r.names)
```
