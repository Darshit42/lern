## RGB Image Representation
In a color image, each pixel is represented by three numbers (Red, Green, Blue). Here we’ll create a tiny 3×3 RGB image and visualize both the image and its numerical representation.

```python
import numpy as np
import matplotlib.pyplot as plt

# Create a tiny 3x3 RGB image
img_rgb = np.array([
    [[255, 0, 0], [0, 255, 0], [0, 0, 255]],   # Red, Green, Blue
    [[255, 255, 0], [0, 255, 255], [255, 0, 255]], # Yellow, Cyan, Magenta
    [[128, 128, 128], [255, 255, 255], [0, 0, 0]]  # Gray, White, Black
], dtype=np.uint8)

# Plot the image and its matrix values
fig, ax = plt.subplots(1, 2, figsize=(8,4))

# Show image
ax[0].imshow(img_rgb)
ax[0].set_title("RGB Image")
ax[0].axis('off')

# Show matrix values
ax[1].imshow(img_rgb)
for i in range(img_rgb.shape[0]):
    for j in range(img_rgb.shape[1]):
        ax[1].text(j, i, str(list(img_rgb[i, j])), ha='center', va='center', color='white')
ax[1].set_title("Pixel RGB Values")
ax[1].axis('off')

plt.show()
```
