```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

# Load grayscale image
img = cv2.imread('sample.jpg', 0)   # replace with your image file

plt.imshow(img, cmap='gray')
plt.title("Original Grayscale Image")
plt.axis('off')
plt.show()
```

## Structuring Elements
- Small binary matrix used to probe the image.
- Shape: rectangle, ellipse, cross.
- Size determines effect of operation.

```python
kernel = np.ones((5,5), np.uint8)  # 5x5 square kernel
print(kernel)
```

## Dilation
- Expands white regions.
- Useful for connecting broken parts of objects.

```python
dilated = cv2.dilate(img, kernel, iterations=5)

plt.imshow(dilated, cmap='gray')
plt.title("Dilated Image")
plt.axis('off')
plt.show()
```

## Erosion
- Shrinks white regions.
- Useful for removing noise.

```python
eroded = cv2.erode(img, kernel, iterations=1)

plt.imshow(eroded, cmap='gray')
plt.title("Eroded Image")
plt.axis('off')
plt.show()
```

## Opening
- Erosion followed by dilation.
- Removes small noise while preserving shape.

```python
opened = cv2.morphologyEx(img, cv2.MORPH_OPEN, kernel)

plt.imshow(opened, cmap='gray')
plt.title("Opening")
plt.axis('off')
plt.show()
```

## Closing
- Dilation followed by erosion.
- Fills small holes inside objects.

```python
closed = cv2.morphologyEx(img, cv2.MORPH_CLOSE, kernel)

plt.imshow(closed, cmap='gray')
plt.title("Closing")
plt.axis('off')
plt.show()
```

## Applications
- Document cleaning (remove noise, fill gaps).
- Medical imaging (highlight structures).
- Industrial inspection (detect defects).

## Lab Exercise
- Apply dilation, erosion, opening, and closing.
- Compare outputs side by side.

```python
titles = ['Original', 'Dilated', 'Eroded', 'Opening', 'Closing']
images = [img, dilated, eroded, opened, closed]

for i in range(5):
    plt.subplot(2,3,i+1)
    plt.imshow(images[i], cmap='gray')
    plt.title(titles[i])
    plt.axis('off')
plt.show()
```

## Mini‑Project: Noise Removal in Scanned Documents
- Input: noisy scanned document.
- Output: cleaned binary version using opening & closing.

```python
doc = cv2.imread('sample.jpg', 0)  # replace with your document image
kernel = np.ones((3,3), np.uint8)

doc_open = cv2.morphologyEx(doc, cv2.MORPH_OPEN, kernel)
doc_clean = cv2.morphologyEx(doc_open, cv2.MORPH_CLOSE, kernel)

plt.imshow(doc_clean, cmap='gray')
plt.title("Cleaned Document")
plt.axis('off')
plt.show()
```
